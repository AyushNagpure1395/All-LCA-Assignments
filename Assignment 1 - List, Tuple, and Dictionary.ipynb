```python
# Creating and displaying a dictionary
my_dict = {"Name": "Ayush Nagpure", "Age": 18, "Course": "Computer Science AI/DS", "Marks": 98, "City": "Pune"}
print(my_dict)

# Adding a new key-value pair to the dictionary
my_dict["Grade"] = "A+"
print(my_dict)

# Deleting a key-value pair using the 'del' statement
del my_dict["City"]
print(my_dict)
```

```python
# Standard dictionary creation using zip() and dict()
list1 = ["Name", "Panel", "Roll No"]
list2 = ["ABC", "B", 34]

# Pair elements from list1 (keys) and list2 (values) into a dictionary
my_dict = dict(zip(list1, list2))
print(my_dict)
```

```python
# Defining the initial dictionary with mixed data types
my_dict = {"Name": "ABC", "Panel": "B", "Roll No": 34, "Marks": [65, 87, 67, 94]}
sorted_keys = sorted(my_dict)

# Extracting and sorting dictionary keys
print(my_dict)
print(sorted_keys)

# Sorting dictionary key-value items
sorted_items = sorted(my_dict.items())
print(sorted_items)
```

```python
# Defining the initial dictionary
my_dict = {"Name": "ABC", "Panel": "B", "Roll No": 34, "Marks": [65, 87, 67, 94]}
print(my_dict)

# Extracting keys and values into separate lists
key_list = list(my_dict.keys())
value_list = list(my_dict.values())

print(key_list)
print(value_list)
```

```python
# Defining the dictionary with numeric data
my_dict = {"Marks 1": 23, "Marks 2": 123, "Marks 3": 43, "Marks 4": 13, "Marks 5": 39}
print(my_dict)

# Calculating sum, count, and mean from dictionary values
total = sum(my_dict.values())
count = len(my_dict.values())
mean_value = total / count

print("Sum of Values:", total)
print("Number of Values:", count)
print("Mean of Values:", mean_value)
```

```python
# Defining a dictionary containing list values
my_dict = {"Name": ["Yash", "Neel", "Dev"], "Roll No": [11, 12, 13], "Marks": [78, 56, 98]}
print(my_dict)

# Searching for elements within specific list values
if "Dev" in my_dict["Name"]:
    print("Name found:", "Dev")
else:
    print("Dev not found")
print()

if 12 in my_dict["Roll No"]:
    print("Roll No found:", 12)
else:
    print("Roll No 12 not found")
print()

# Finding maximum value from a nested list
greatest_marks = max(my_dict["Marks"])
print("Greatest Marks:", greatest_marks)
```

```python
# Standard loop implementation with conditional logic
my_string = "Programming"

freq = {}

for char in my_string:
    if char.isalpha():                 # consider only letters
        char = char.lower()            # treat 'P' and 'p' as same
        if char in freq:
            freq[char] += 1
        else:
            freq[char] = 1

print("Original String:", my_string)
print("Letter Frequency:")
print(freq)
```

```python
# Standard floor division and modulo operations
numerator = int(input("Enter the numerator: "))
denominator = int(input("Enter the denominator: "))

quotient = numerator // denominator
remainder = numerator % denominator

result = (quotient, remainder)

print("Quotient and Remainder as Tuple:", result)
```

```python
# Standard implementation reading user input into date tuples
from datetime import date

d1 = int(input("Enter day for Date 1: "))
m1 = int(input("Enter month for Date 1: "))
y1 = int(input("Enter year for Date 1: "))
date1_tuple = (d1, m1, y1)

d2 = int(input("Enter day for Date 2: "))
m2 = int(input("Enter month for Date 2: "))
y2 = int(input("Enter year for Date 2: "))
date2_tuple = (d2, m2, y2)

# Instantiate date objects using tuple indexing (Year, Month, Day)
date1 = date(date1_tuple[2], date1_tuple[1], date1_tuple[0])   # y, m, d
date2 = date(date2_tuple[2], date2_tuple[1], date2_tuple[0])

difference = date2 - date1

print("Date 1:", date1_tuple)
print("Date 2:", date2_tuple)
print("Number of days between the two dates:", abs(difference.days))
```
