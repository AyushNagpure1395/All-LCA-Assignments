```python
# Define three numbers
a = 45
b = 89
c = 12

# Determine the largest using conditional logic
if a >= b and a >= c:
    largest = a
elif b >= a and b >= c:
    largest = b
else:
    largest = c

print(f"The largest number is: {largest}")
```

```python
# The 'and' operator
age = 22
has_license = True

# Returns True only because BOTH conditions evaluate to True
can_drive = (age >= 18) and has_license
print("Can drive:", can_drive)  # Output: True

# One condition is False (age < 18)
is_minor_with_license = (age < 18) and has_license
print("Minor with license:", is_minor_with_license)
```

```python
# The 'or' operator
is_weekend = False
is_holiday = True

# Returns True because AT LEAST ONE condition is True
can_relax = is_weekend or is_holiday
print("Can relax today:", can_relax)  # Output: True

# Both conditions are False
is_work_day = (not is_weekend) and (not is_holiday)
print("Is regular work day:", is_work_day)
```

```python
# The 'not' operator
is_logged_in = False

# Reverses the boolean value from False to True
if not is_logged_in:
    print("Access Denied: Please log in to continue.")
```
