```python
def is_right_triangle(a, b, c):
    s = sorted([a, b, c])
    return s[0]**2 + s[1]**2 == s[2]**2

# User inputs
x = float(input("Enter side 1: "))
y = float(input("Enter side 2: "))
z = float(input("Enter side 3: "))

# Result
if is_right_triangle(x, y, z):
    print("Yes, it is a right-angled triangle.")
else:
    print("No, it is not a right-angled triangle.")
```
