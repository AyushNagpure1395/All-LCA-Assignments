```python
# Create two matrices using nested lists
matrix1 = [[1, 2, 3],
           [4, 5, 6]]

matrix2 = [[7, 8, 9],
           [10, 11, 12]]

# Initialize a result matrix with the same size (2x3)
result = [[0, 0, 0],
          [0, 0, 0]]

# Perform matrix addition using nested loops
for i in range(len(matrix1)):                         #Loop through rows
  for j in range(len(matrix1[0])):                    #Loop through columns
    result[i][j] = matrix1[i][j] + matrix2[i][j]

# Display the result
print("Matrix 1:")
for row in matrix1:
    print(row)
print("Matrix 2:")
for row in matrix2:
    print(row)

print("Addition of Matrices:")
for row in result:
  print(row)
```
